- ## Specialised arrays & some special functions for changing array shape or creating arrays with randomised numbers:
- ##### np.ones((shape)): Creates arrays of all 1s of the required shape. 
- ##### np.zeros((shape)): Creates arrays of all 0s of the required shape.
- ##### np.random.randint(Start,End,size=()): Creates randomised arrays of integers of the required shape/size with start,end(excluded).
- ##### np.random.random((shape)) or np.random.rand(shape): Creates randomised arrays of floats of the required shape.
- ##### Arr.reshape(shape): Changes the shape of the existing arrays to the required shape.
- ##### np.arange(Start,Stop,Step): Creates arrays of values from start to end(stop:excluded).
- ##### np.random.seed(Any Int): Sets the seed for reproducibility of random numbers to generate the same set of numbers everytime the code is runned.

- ## Image Storing & Processing:
- ##### (8-bit)Images: Images are readed by imread() method of matplotlib's image module & stored as 3D matrix, in which each dimension is again a 2D matrix itself.
- ##### To display the images Image() method is used, which can be imported from IPython's display module.
- ##### The three dimensions are (Height X Width X Channel),i.e., (rows(Y),columns(X),RGB channels).
- ##### Total Pixels=Height X Width & Total Values = Total Pixels X Channels(3).
- ##### Pixel Values(Colour Intensity): 0-255
- ##### 0: Darkest(no color).
- ##### 255: Lightest(full color).
- ##### RGB[:,:,0/1/2] : shape (rows,columns): 2D matrix where each value represents how much color(RGB) intensity is at that pixel.
- ##### Pixel at position (row=0,column=0):
- ##### RGB Pixel=Red Matrix+Green Matrix+Blue Matrix
- ##### Examples: Red Channel = RGB[:,:,0]-All red channel. 
- ##### Green Channel = RGB[:,:,1]-All green channel.
- ##### Blue Channel = RGB[:,:,2]-All blue channel.

```python
from matplotlib.image import imread
from IPython.display import Image
```

```python
Image('Cat.png')
```

```python
Cat=imread('Cat.png')
```

```python
Red_Channel=Cat[:,:,0]
Red_Channel
```
