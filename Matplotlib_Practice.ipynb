```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

>- ## Two Types Of Plotting Approaches:
1) #### Pyplot Module(MATLAB-style): Simple & quick plots with limited control over the complex layouts.Mostly used for simple exploratory data analysis.
2) #### Object Oriented Interface(OOP): Detailed & complex plots with more control over the complex layouts.Mostly used for professional,public plots,where the code can be reused.

- #### Pyplot Module Approach Syntax:
- ***plt.plot()***:
- Creates interactive plotting with a global state.
- Global State: Creates visualiztion (figure and axes) of the current available data without us explicitly referencing them.
- ***.plot() parameters:***
- `x=`: Selects the data for x-axis.
- `y=`: Selects the data for y-axis.
- `c(color)=`: Finds the min and max values in a column and maps each values to a certain color(Numerical Columns).
-  For categorical columns maps each category to a distinct color. 
- `kind='line'`: Controls the type of graph to be plotted.

```python
X=np.array([1,2,3,4,5])
Y=np.array([2,4,6,8,10])
plt.plot(X,Y) # Function that creates a graph/plot.
plt.xlabel("X-Axis") # Set x label.
plt.ylabel("Y-Axis") # Set y label.
plt.title("Quick Plot") # Set the title of the plot.
plt.show() # Display the plot.
```

```python
Data = {
    'Name':['Ali','Umar','Afaq','Mubashir','Manan'],
    'Age':[20,21,18,14,16],
    'Class':[12,13,10,6,9]
}
DF=pd.DataFrame(Data)
DF
```

```python
DF.plot(kind='scatter',x='Name',y='Class',c='Age')
plt.title("Person Class Distribution By Their Age")
plt.show()
```

- #### Object Oriented Approach Syntax:
- ***fig,ax=plt.sublots()***: Creates a figure and axes for plotting.
- ***.sublots() parameters:***
- `nrows=int & ncols=int`: These control the adjustment/layout of comparison subplots,which determines how comparison is done.
- Different values of nrows & ncols represent different order of comparisons:
- `nrows=1,ncols=1`: Standard single plot.
- `nrows=1,ncols=2`: Side by side comparison of two plots horizontally.One row & two columns of subplots.
- `nrows=2,nrows=1`: (Stacked Vertically) Vertical comparison of data on the same x-axis.
- `nrows=2,ncols=2`: Four plots grid. (2X2)
- `nrows=3,ncols=3`: Nine plots grid. (3X3)
- `nrows=1,ncols=3`: 3 subplots in one row.
- `nrows=3,ncols=1`: 3 subplots in one column.
- `figsize=(Height,Width)`: Controls figure size.
- `facecolor='color'`: Controls the background color.
- `edgecolor='color'`: Controls the edge color.
- `dpi=int`: Controls the resolution of the plotting.
- `constrained_layout=True/False`: Controls the layout(Mutiple Axes/Plots) adjustment in a figure.
- `tight_layout=True/False`:Controls the layout space adjustment.
- ***ax.plot() parameters:***
- `X,Y`: Data to be plotted in the graph.
- `label='title'`: Label of the plot.
- `color='color'`: Color of the plot line.
- `linestyle='-/--/:'`: Controls the type of line plotted.
- `linewidth=int`: Controls the width of the line.
- `marker='o/D/^/s/*/+'`: Markers are indicators or symbols that identify certain data values.This selects the symbols for identification.
- `markersize=int`: Sets the size of the marker points.
- `alpha=0-1`: Controls the transparency/opacity of the plots.
- ***ax.set() parameters:***
- `title='ABC'`: Sets the main title of the entire plots.
- `xlabel='X-axis'`: Sets the x axis labels.
- `ylabel='Y-axis'`: Sets the y axis labels.
- These can also be set individually:
- `ax.set_title('Title',fontsize=int,fontweight='bold')`
- `ax.set_xlabel('X')`
- `ax.set_ylabel('Y')`
- ***ax.legend()***:
- Creates a reference table with colors,linestyles,markers,etc., representing specific data points.
- It contains handles & labels.
- ***Handles:*** The visual elements(color patches,markers,lines).
- ***Labels:*** The text/name of each visual element.
- ***ax.legend() parameters:***
- `ax.legend(['Label1','Label2'])`
- `loc='upper left/upper right/bottom left/bottom right'`
- `title='Legend Title'`
- ***Grid Lines:***
- `ax.grid(True,alpha=0-1,linestyle='-')`
- ***Limits & Ticks(Intervals):***
- `ax.set_xlim([0,10])`: Sets the limit of the x-axis values.
- `ax.set_ylim([0,10])`: Sets the limit of the y-axis values.
- `ax.set_xticks([0,2,4,6,7,10])`: Sets the intervals of the x-axis.

```python
fig0,ax0 = plt.subplots(figsize=(10,5),
                      facecolor='lightblue',
                      edgecolor='yellow',
                     dpi=200)
ax0.plot(X,Y,label='X/Y Datapoints',color='green',linestyle='-',
        linewidth=2,marker='o',markersize=5,alpha=0.2)
ax0.set(title='OOP Plotting',xlabel="X-Axis",ylabel="Y-Axis")
ax0.legend(loc='upper left')
plt.plot()
fig1,ax1 = plt.subplots(figsize=(10,5),
                      facecolor='skyblue',
                      edgecolor='yellow',
                     dpi=200)
ax1.plot(X,Y,label='X/Y Datapoints',color='blue',linestyle='--',
        linewidth=2,marker='s',markersize=5,alpha=0.2)
ax1.set(title='OOP Plotting',xlabel="X-Axis",ylabel="Y-Axis")
ax1.legend(loc='lower right')
plt.plot()
fig2,ax2 = plt.subplots(figsize=(10,5),
                      facecolor='yellow',
                      edgecolor='blue',
                     dpi=200)
ax2.plot(X,Y,label='X/Y Datapoints',color='red',linestyle=':',
        linewidth=2,marker='D',markersize=5,alpha=0.5)
ax2.set(title='OOP Plotting',xlabel="X-Axis",ylabel="Y-Axis")
ax2.grid(True,alpha=0.2,linestyle='-')
ax2.legend() # This function gets the tuple of handles and labels automatically.
plt.plot()
```

```python
Arr=np.linspace(1,10,20) # (Start,Stop(incuded),Step) Creating array of equally spaced values determined by step.
fig3,ax3=plt.subplots(figsize=(20,10),facecolor='lightgreen',dpi=200)
ax3.plot(Arr,label='Randomized Numbers',color='blue',linestyle='-',linewidth=3,marker='D',markersize=5,alpha=1)
ax3.set_xlabel('X-Axis')
ax3.set_ylabel('Y-Axis')
ax3.set_title("Randomized Array",fontsize=14,fontweight='bold')
ax3.set_xlim([0,20])
ax3.set_ylim([0,11])
ax3.set_xticks([0,1.5,3.0,4.5,6.0,7.5,9.0,10.5,12.0,13.5,15.0,16.5,18.0,19.5])
ax3.grid(True,alpha=0.8,linestyle='--')
ax3.legend(loc='upper left')
plt.show()
```

- ***Scatter Plot:*** This plot is used for:
- Showing relationship or correlation between two numerical valriables.
- Detecting outliers
- Identifying clusters and groupings

- Showing distribution of two variables together

```python
Data2={
    'Force(N)':[10,20,30,40,10,20,30,40,50,50,60,60],
    'Mass(kg)':[3.5,3.5,3.5,3.5,5.5,5.5,5.5,5.5,3.5,5.5,3.5,5.5],
    'Acceleration(m/s^2)':[2.85,5.71,8.57,11.42,1.82,3.64,5.45,7.27,14.28,9.09,17.14,10.09]
}
DF2=pd.DataFrame(Data2)
DF2
```

```python
DF2.plot(kind='scatter',x='Acceleration(m/s^2)',y='Force(N)',c='Mass(kg)',marker='D',alpha=1)
plt.xlabel('Acceleration in m/s^2')
plt.ylabel('Force Applied in Newtons(N)')
plt.title("Force Mass Acceleration (Newton's Second Law)")
plt.grid(True,alpha=0.5)
plt.show()
```

- ***.scatter() Parameters:***
- `c(color)=DF['col']`: Finds the min and max values in a column and maps each values to a certain color(Numerical Columns).
-  For categorical columns maps each category to a distinct color.
-  `cmap(color map)='color'`: Chooses color for mapping.
-  For creating color bars for showing how values are mapped to different color intensities:
-  `s(Markersize)=int`
- `plt.colorbar(Scatter,label='L')`
- As legend stores handles and labels automatically and shows them as reference table in the plot, to unpack tuples of handles and labels in case of scatter plots, unpack operator `*` is used and for getting the individual handles and labels `.legend_elements()` function is used.
- For creating a horizontal mean reference line,showing how much data is scattered around the mean.
- `ax.axhline(Nom_col.mean(),linestyle='--')`

- ***Histogram:*** It is used for showing distribution(normal,skewed,etc) of a single continuous(numerical) variable by dividing it into bins and counting frequencies.
- It is used for:
- Understanding data distribution
- Identifying outliers
- Finding data density and spread
- Quality control and process monitoring
- Initial data exploration
- ***ax.hist() Parameters:***

- `label='L'`

- `color='color'`
- `orientation='vertical'/'horizontal'`
- `density=True/False`: Probability density plot: Showing probability at a certain value of the variable.
- `commulative=True/False`: Shows commulative(Total) frequency distribution of values.
- `data=DF`: For DataFrame data.
- `align='mid'`: Alignment of bins.
- `stacked=True/False`: Stack histogram.
- `histtype='bar'`: For selecting type of histogram.
- `bottom=0`: Starting area of bins.
- `weight=None`: Weights for each datapoint.
- `range=None`: Min.Max for binning.

- ***Box Plot:*** This plot is used for showing data distribution through quartiles(IQR Box:Q1 & Q3),median(Middle line in the box),and outliers(Points) in a compact format with a range of data(Whiskers).
- Compare Distributions accross different groups
- Any data with different groups to compare
- Outlier Detection
- Statistical Summary(Q1,Q3,Median,Outliers,Range(Whiskers))

```python
ax.boxplot?
```

- ***ax.boxplot() Parameters:***
- `x=Data(Array-Like)`
- `sym='+'`: Symbol for outliers
- `vert=True/False`: Vertical box or horizontal box
- `whis=1.5`: Whisker(Range) length (IQr multiplier)
- `positions=None`: Position of boxes
- `patch_artist=True`: Addiing colors to the plot 
- `widths=0-1`: Width of boxes
- `showmeans=True`: Show mean
- `showcaps=True`: Show caps on whiskers
- `showbox=True`: Show box
- `showfliers=True`: Show outliers
- `data=None`: Dataframe for data mapping.
- `labels=None`: Labels dor each box.
- `flierprops=dict(marker='s',markersize=int,markerfacecolor='red',alpha=0-1)`: Controlling properties of outlier through dict(Dictionary) contructor.
- `capprops=None(Default)=dict()`
- `boxprops=None=dict()`
- `whiskersprops=None=dict()`
- `meanprops=None=dict()`

```python
X1=np.random.randint(1,10,20)
Outliers=np.array([-15,-20,-12,0,20,25,15,12,16])
X2=np.random.normal(1,20,size=20)
Combined_Arr=np.concatenate((X1,Outliers))
fig,(ax0,ax1)=plt.subplots(nrows=1,ncols=2,figsize=(25,10),
                           dpi=200,constrained_layout=True)
ax0.boxplot(Combined_Arr,vert=False,widths=0.5,
            showmeans=True,showcaps=True,showfliers=True,
            sym='o',showbox=True,patch_artist=True
           )
ax0.set_title("Uniform Values Distribution",fontsize=15,
              fontweight='bold')
ax0.set_xlabel("Random Array Values",fontsize=15)
ax0.set_ylabel('Distributions',fontsize=15)
ax1.boxplot(x=X2,vert=True,widths=0.5,
            showmeans=True,showcaps=True,showfliers=True,
            sym='D',showbox=True,patch_artist=True
           )
ax1.set_title("Normal Values Distribution",fontsize=15,
              fontweight='bold')
ax1.set_xlabel("Distributions",fontsize=15)
ax1.set_ylabel('Random Array Normal Values',fontsize=15)
plt.show()
```
