- Getting the data ready for model training
- Preprocessing
- Handling Missing Values(Imputation)

```python
X=DF.drop('Price',axis=1)
Y=DF['Price']
```

```python
X
```

```python
Y
```

```python
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
Num_Cols=["Odometer (KM)","Doors"]
Cat_Cols=DF.select_dtypes(include='object').columns
Cat_Imputer=SimpleImputer(strategy='constant',fill_value='missing')
Num_Imputer=SimpleImputer(strategy='mean')
Imputer=ColumnTransformer(
    [("Category Imputing",Cat_Imputer,Cat_Cols),
    ("Numerical Imputing",Num_Imputer,Num_Cols)])
Filled_X=Imputer.fit_transform(X)
```

```python
Cat_Cols
```

```python
Filled_X
```

```python
Car_Filled_DF=pd.DataFrame(Filled_X,columns=['Make','Colour','Odometer (KM)','Doors'])
Car_Filled_DF
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
Cat_Cols2=['Make','Colour']
One_Hot=OneHotEncoder()
Encoder=ColumnTransformer(
    [("One_Hot_Encoding",One_Hot,Cat_Cols2)],remainder="passthrough")
X_Transformed=Encoder.fit_transform(Car_Filled_DF)
```

```python
X_Transformed
```

```python
X_Transformed.shape , X_Transformed.ndim
```

```python
print(type(X_Transformed))
```

```python
X_DF=pd.DataFrame( X_Transformed , 
                   columns=Encoder.get_feature_names_out())
X_DF
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
RFR=RandomForestRegressor()
X_Train,X_Test,Y_Train,Y_Test = train_test_split(X_Transformed,Y,test_size=0.2,random_state=42)
RFR.fit(X_Train,Y_Train)
```

```python
RFR.fit(X_Train,Y_Train);
```

```python
Y_Pred=RFR.predict(X_Test)
Y_Pred
```

```python
Y_Test
```

```python
RFR.get_params()
```

```python
from sklearn.metrics import r2_score
R2=r2_score(Y_Pred,Y_Test)
R2
```

```python
from sklearn.model_selection import cross_val_score
Cross_Validations=cross_val_score(RFR,X_Transformed,Y,cv=10)
```

```python
Cross_Validations
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import RandomizedSearchCV
Parameter_Grid1 = {
    'n_estimators':[100,200],
    'max_depth': [3,5,10]
}
Parameter_Grid2 = {
     'n_estimators':[100,200,300,500],
    'max_depth': [5,10,20],
    'min_samples_split': [2,10,15]
}
```

```python
Grid_Search=GridSearchCV(
    RandomForestRegressor(random_state=42),
    Parameter_Grid1,
    n_jobs=1,
    cv=10,
    scoring='r2',
    verbose=2
)
```

```python
Random_Search=RandomizedSearchCV(
    RandomForestRegressor(random_state=42),
    Parameter_Grid2,
    n_iter=20,
    n_jobs=1,
    random_state=42,
    scoring='r2',
    verbose=2
)
```
